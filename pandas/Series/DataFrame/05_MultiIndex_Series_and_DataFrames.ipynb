Series is 1D and DataFrames are 2D objects

But why?

And what exactly is index?


```python
# can we have multiple index? Let's try
index_val= [('cse',2019),('cse',2020),('cse',2021),('cse',2022),('ece',2019),('ece',2020),('ece',2021),('ece',2022)]
a = pd.Series([1,2,3,4,5,6,7,8],index = index_val)
a
# it is possible but not correct working
```

```python
# how to create multiindex object
# 1. pd.MultiIndex.from_tuples()
index_val=[('cse',2019),('cse',2020),('cse',2021),('cse',2022),('ece',2019),('ece',2020),('ece',2021),('ece',2022)]
multiindex = pd.MultiIndex.from_tuples(index_val)
multiindex.levels[0]  # -> Shows first cols of index
multiindex.levels[1]  # -> Shows second cols of index

# 2. pd.MultiIndex.from_product()    it's like Herarchical clustering
pd.MultiIndex.from_product([['cse','ece'],[2019,2020,2021,2022]]) # -> cse * second_list & ece* second_list
```

```python
# creating inside multiindex object
```

```python
# multiindex df from columns perspective
branch_df2 = pd.DataFrame(
    [
        [1,2,0,0],
        [3,4,0,0],
        [5,6,0,0],
        [7,8,0,0],
    ],
    index = [2019,2020,2021,2022],
    columns = pd.MultiIndex.from_product([['delhi','mumbai'],['avg_package','student']])
)
branch_df2
```

```python
branch_df2.loc[2019]
```

```python
# multiindex df in terms of both cols and index
branch_df3 = pd.DataFrame(
    [
[1,2,0,0],
[3,4,0,0],
[5,6,0,0],
[7,8,0,0],
[9,10,0,0],
[11,12,0,0],
[13,14,0,0],
[15,16,0,0],
    ],
index = multiindex,
columns = pd.MultiIndex.from_product([['delhi','mumbai'],['avg_package','students']])
)
branch_df3
```

# Stacking and Unstacking

```python
branch_df3.stack()
```

```python
branch_df3.unstack()
```

## Working with  multiindex dataframes

```python
# head and tail
branch_df3.head()
# tail
branch_df3.tail()
# shape
branch_df3.shape
# info
branch_df3.info()
# duplicated -> isnull
branch_df3.duplicated()
branch_df3.isnull()
```

```python
# Extracting rows single
branch_df3.loc[('cse',2022)]
```

```python
# Multiple
branch_df3.loc[('cse',2019):('ece',2022):2]
```

```python
# using iloc
branch_df3.iloc[0:5:2]
```

```python
# extracting cols
branch_df3['delhi']['students']
```

```python
branch_df3.iloc[:,1:3]
```

```python
# Extracting both
branch_df3.iloc[[0,4],[1,2]]
```

```python
# sort index
# both -> descending -> diff order
# based on one level
branch_df3.sort_index(ascending=False)
branch_df3.sort_index(ascending=[False,True])
```

```python
# multiindex dataframe(col)  -> transpose
branch_df3.transpose()
```

```python
# swaplevel
branch_df3.swaplevel(axis=1)
```
