```python
# looping on groups
# find the highest rated movie of each genre
df = pd.DataFrame(columns=movies.columns)
for group,data in genres:
  df =  pd.concat([df, data[data['IMDB_Rating']== data['IMDB_Rating'].max()]], ignore_index=True)

df.head(5)
```

```python
#split (apply) combine
# apply -> builtin function
genres.apply(min).head(4)
```

```python
# find number of movies starting with A for each group
def foo(group):
  return group['Series_Title'].str.startswith('A').sum()
```

```python
genres.apply(foo)
```

```python
# find ranking of each movie in the group according to IMDB score
def rank_movie(group):
  group['genre_rank'] = group['IMDB_Rating'].rank(ascending= False)
  return group
```

```python
genres.apply(rank_movie)
```

```python
# find normalization IMDB rating group wise
# Normalization Formula: x_norm = (x - x_min) / (x_max - x_min)′    -> x = normalize_rating
def normalize_rating(group):
  group['normalize_rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min())/ (group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
  return group
```

```python
genres.apply(normalize_rating)
```

```python
ipl = pd.read_csv('/content/deliveries.csv')
ipl.head()
```

```python
ipl.shape
```

```python
# find the top 10 batsman in terms of runs
ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(10)
```

```python
# find the batsman with max no of sixes
six = ipl[ipl['batsman_runs'] == 6]
six.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(1)
```

```python
# find batsman with most number of 4's and 6's runs in last 5 overs
temp_df = ipl[ipl['over'] > 15]
temp_df = temp_df[(temp_df[('batsman_runs')] == 4) | (temp_df[('batsman_runs')] == 6)]
temp_df.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(1)
```

```python
# find V kohli's record against all teams
temp_df = ipl[ipl['batsman'] == 'V Kohli']
temp_df.groupby('bowling_team')['batsman_runs'].sum().reset_index()
```

```python
# create a function that can return the highest score of any batsman
temp_df = ipl[ipl['batsman'] == 'V kohli']
temp_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1)
```

```python
def highest_score(batsman):
  temp_df = ipl[ipl['batsman'] == batsman]
  return temp_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values
```

```python
highest_score('CH Gayle')
```
