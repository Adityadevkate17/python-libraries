```python
# Import the library
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# from a pandas dataFrame
batsman = pd.read_csv('/content/sharma-kohli.csv')
batsman

plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
batsman
```

```python
# ploting multiple plots
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# labels title
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title('Rohit sharma VS virat kohali career Comprison')
plt.xlabel('season')
plt.ylabel('Runs')
```

```python
# colors(hex)
plt.plot(batsman['index'],batsman['V Kohli'],color='#F08080')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#180501')

plt.title('Rohit sharma VS virat kohali career Comprison')
plt.xlabel('season')
plt.ylabel('Runs')
```

```python
# line(width and style)
plt.plot(batsman['index'],batsman['V Kohli'],color='#F08080',linestyle='solid',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'],color='#180501',linestyle='dotted',linewidth=2)

plt.title('Rohit sharma VS virat kohali career Comprison')
plt.xlabel('season')
plt.ylabel('Runs')
```

```python
 # marker(size)
plt.plot(batsman['index'],batsman['V Kohli'],color='#F08080',linestyle='solid',linewidth=3,marker='D',markersize=8)
plt.plot(batsman['index'],batsman['RG Sharma'],color='#180501',linestyle='dotted',linewidth=2)

plt.title('Rohit sharma VS virat kohali career Comprison')
plt.xlabel('season')
plt.ylabel('Runs')
```

```python
# legend -> location
plt.plot(batsman['index'],batsman['V Kohli'],color='#F08080',linestyle='solid',linewidth=3,marker='D',markersize=8,label='virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#180501',linestyle='dotted',linewidth=2,marker='*',markersize=5,label='sharma')

plt.title('Rohit sharma VS virat kohali career Comprison')
plt.xlabel('season')
plt.ylabel('Runs')

plt.legend(loc='best')
```

```python
# limiting code
price = [48000,53000,55000,49000,4500000]
year = [2015,2016,2017,2018,2019]

plt.plot(year,price)
plt.ylim(0,175000)
plt.xlim(2017,2019)
```

# Scatter Plots

Bivariate Analysis

numerical vs numerical

use case - finding correlation


```python
# plt.scatter simple function
x = np.linspace(-10,10,50)
x
y = 10*x + 3 + np.random.randint(0,300,50)

plt.scatter(x,y)
```

```python
# plt.scatter on pandas data
df = pd.read_csv('/content/batter.csv')
df = df.head(50)
df
```

```python
plt.scatter(df['avg'],df['strike_rate'],color='red',marker='*')
plt.title('Avg and SR analysis of top 50 Btasman')
plt.xlabel('Average')
plt.ylabel('SR')
```

# Bar Chart
· Bivariate Analysis

. Numerical vs Categorical.

· Use case - Aggregate analysis of groups

```python
# simple bar chart
children = [10,20,30,40,10,30]
colors = ['red','blue','green','yellow','pink','purple']

plt.bar(colors,children)
```

```python
# horizontal bar chart
plt.barh(colors,children,color='black')
```

```python
# color and label
df = pd.read_csv('/content/batsman_season_record.csv')
df
```

```python
plt.bar(np.arange(df.shape[0])- 0.2,df['2015'],width=0.2)
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2)
plt.bar(np.arange(df.shape[0])+ 0.2,df['2017'],width=0.2)

plt.xticks(np.arange(df.shape[0]), df['batsman'])

plt.show()
```

```python
np.arange(df.shape[0])
```

```python
# a problem
children = [10,20,40,10,30]
colors = ['red red red red red red', 'blue blue blue blue', 'green green green green green', 'yellow yellow yellow yellow ', 'pink pinkpinkpink']

plt.bar(colors, children)
plt.xticks(rotation='vertical')
```

```python
# Stacked Bar chart
plt.bar(df['batsman'], df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'], bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=(df['2016'] + df['2017']),label='2015')

plt.legend()
plt.show()
```

# Pie Chart

· Univariate/Bivariate Analysis

· Categorical vs numerical

. Use case - To find contibution on a standard scale

```python
# simple data
data = [23,45,100,20,49]
subjects = ['eng','science','maths','sst','hindi']
plt.pie(data,labels=subjects)
plt.show()
```

```python
# dataset
df = pd.read_csv('/content/gayle-175.csv')
df
```

```python
# percentage
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%')
plt.show()
```

```python
#colors
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',colors=['blue','red','green','yellow','pink','brown'])
plt.show()
```

```python
plt.figure(figsize=(8, 8))
```

```python
# explode show
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.2,0,0,0,0,0],shadow=True)
plt.show()
```
