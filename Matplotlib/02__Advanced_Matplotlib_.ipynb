```python
sample_df
```

```python
x = [0,1,5,25]
y = [0,10,13,0]
z = [0,13,20,9]

fig = plt.figure(figsize=(6,8))

ax = plt.subplot(projection='3d')

ax.scatter3D(x,y,z,s=[100,100,100,100])
ax.plot3D(x,y,z,color='red')
```

# 3D surface plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
x
```

```python
xx, yy = np.meshgrid(x,y)
xx.shape
```

```python
z = xx**2 + yy**2
z.shape
```

```python
fig = plt.figure(figsize=(10,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)
fig = plt.figure(figsize=(10,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')

fig.colorbar(p)
```

# contour plots

```python
fig = plt.figure(figsize=(10,8))

ax = plt.subplot()

p = ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

# Heatmap

```python
delivery = pd.read_csv('/content/IPL_Ball_by_Ball_2008_2022.csv', engine='python', on_bad_lines='skip')
delivery.head()
```

```python
temp_df = delivery[(delivery['ballnumber'].isin([1,2,3,4,5,6]) & (delivery['batsman_run']==6))]
```

```python
grid = temp_df.pivot_table(index='overs',columns='ballnumber',values='batsman_run',aggfunc='count')
```

```python
plt.figure(figsize=(10,8))
plt.imshow(grid)
plt.yticks(np.arange(len(grid.index)), [f'{int(o)+1}' for o in grid.index])
plt.xticks(np.arange(len(grid.columns)), grid.columns)
plt.colorbar()
```
